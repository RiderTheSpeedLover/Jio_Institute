```python
# Import necessary libraries
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder

# Step 1: Upload CSV File
# For local files, use the path to your uploaded CSV
file_path = '/content/market basket analysis.csv'  # Replace with the actual path of your file
data = pd.read_csv(file_path, header=None)

# Step 2: Prepare the Data (Converting the data into transactions format)
# Convert the data into a list of transactions (each row as a transaction)
transactions = data.apply(lambda row: row.dropna().tolist(), axis=1).tolist()

# Step 3: Convert data into one-hot encoded format using TransactionEncoder
te = TransactionEncoder()
te_data = te.fit(transactions).transform(transactions)
df = pd.DataFrame(te_data, columns=te.columns_)

# Step 4: Apply the Apriori algorithm with a lower support threshold
# Try lowering the min_support to capture more frequent itemsets
min_support_value = 0.05  # Lowered to 5%

# Apply the Apriori algorithm
frequent_itemsets = apriori(df, min_support=min_support_value, use_colnames=True)

# Print frequent itemsets to inspect
print("Frequent Itemsets:")
print(frequent_itemsets)

# Step 5: Generate association rules from the frequent itemsets
if not frequent_itemsets.empty:
    # Lower the threshold for lift to 0.5 and try again
    rules = association_rules(frequent_itemsets, metric="lift", min_threshold=0.5)

    # Filter rules to include all possible combinations
    rules = rules[(rules['antecedents'].apply(len) > 0) & (rules['consequents'].apply(len) > 0)]

    # Print the association rules
    print("\nAssociation Rules (Antecedents and Consequents):")
    print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
else:
    print("No frequent itemsets found. Try lowering the support threshold further or check your dataset.")
```
