```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import dendrogram, linkage

# Load dataset
df = pd.read_csv("wine_Dataset.csv")

# Standardize the data (important for clustering)
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)

### 1️⃣ K-MEANS CLUSTERING ###
print("\n===== K-Means Clustering =====")
# Choosing optimal K using Elbow Method
wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(df_scaled)
    wcss.append(kmeans.inertia_)

# Plot Elbow Method
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker="o", linestyle="--")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS (Within-Cluster Sum of Squares)")
plt.title("Elbow Method for Optimal K")
plt.grid()
plt.show()

# Fit KMeans with optimal K (assuming 3 for demonstration)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["KMeans_Cluster"] = kmeans.fit_predict(df_scaled)
print("Cluster Labels Assigned by K-Means:\n", df["KMeans_Cluster"].value_counts())

### 2️⃣ HIERARCHICAL CLUSTERING ###
print("\n===== Hierarchical Clustering =====")
# Perform Agglomerative Clustering
hierarchical = AgglomerativeClustering(n_clusters=3)
df["Hierarchical_Cluster"] = hierarchical.fit_predict(df_scaled)

# Create Dendrogram
plt.figure(figsize=(10, 6))
linked = linkage(df_scaled, method="ward")
dendrogram(linked)
plt.xlabel("Samples")
plt.ylabel("Distance")
plt.title("Hierarchical Clustering Dendrogram")
plt.show()

print("Cluster Labels Assigned by Hierarchical Clustering:\n", df["Hierarchical_Cluster"].value_counts())

### 3️⃣ DENSITY-BASED CLUSTERING (DBSCAN) ###
print("\n===== Density-Based Clustering (DBSCAN) =====")
# Fit DBSCAN
dbscan = DBSCAN(eps=1.5, min_samples=5)
df["DBSCAN_Cluster"] = dbscan.fit_predict(df_scaled)

# Visualize DBSCAN Clusters
sns.scatterplot(x=df["Alcohol"], y=df["Malic"], hue=df["DBSCAN_Cluster"], palette="viridis")
plt.title("DBSCAN Clustering Results")
plt.xlabel("Alcohol")
plt.ylabel("Malic Acid")
plt.show()

print("Cluster Labels Assigned by DBSCAN:\n", df["DBSCAN_Cluster"].value_counts())

# Save results
df.to_csv("wine_clustering_results.csv", index=False)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import dendrogram, linkage

# Load dataset
df = pd.read_csv("wine_Dataset.csv")

# Standardize the data (important for clustering)
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)

### 1️⃣ K-MEANS CLUSTERING ###
print("\n===== K-Means Clustering =====")
# Choosing optimal K using Elbow Method
wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(df_scaled)
    wcss.append(kmeans.inertia_)

# Plot Elbow Method
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker="o", linestyle="--", color="b")
plt.xlabel("Number of Clusters (K)", fontsize=12)
plt.ylabel("WCSS (Within-Cluster Sum of Squares)", fontsize=12)
plt.title("Elbow Method for Optimal K", fontsize=14)
plt.grid()
plt.show()

# Fit KMeans with optimal K (assuming 3 for demonstration)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["KMeans_Cluster"] = kmeans.fit_predict(df_scaled)

# Print K-Means Clustering Results
print("Cluster Labels Assigned by K-Means:\n", df["KMeans_Cluster"].value_counts())

### 2️⃣ HIERARCHICAL CLUSTERING ###
print("\n===== Hierarchical Clustering =====")
# Perform Agglomerative Clustering
hierarchical = AgglomerativeClustering(n_clusters=3)
df["Hierarchical_Cluster"] = hierarchical.fit_predict(df_scaled)

# Create Dendrogram
plt.figure(figsize=(12, 6))
linked = linkage(df_scaled, method="ward")
dendrogram(linked)
plt.xlabel("Samples", fontsize=12)
plt.ylabel("Distance", fontsize=12)
plt.title("Hierarchical Clustering Dendrogram", fontsize=14)
plt.show()

print("Cluster Labels Assigned by Hierarchical Clustering:\n", df["Hierarchical_Cluster"].value_counts())

### 3️⃣ DENSITY-BASED CLUSTERING (DBSCAN) ###
print("\n===== Density-Based Clustering (DBSCAN) =====")
# Fit DBSCAN
dbscan = DBSCAN(eps=1.5, min_samples=5)
df["DBSCAN_Cluster"] = dbscan.fit_predict(df_scaled)

# Visualize DBSCAN Clusters
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df["Alcohol"], y=df["Malic"], hue=df["DBSCAN_Cluster"], palette="viridis", legend="full")
plt.title("DBSCAN Clustering Results", fontsize=14)
plt.xlabel("Alcohol", fontsize=12)
plt.ylabel("Malic Acid", fontsize=12)
plt.legend(title="Cluster Label")
plt.grid()
plt.show()

print("Cluster Labels Assigned by DBSCAN:\n", df["DBSCAN_Cluster"].value_counts())

### 🔹 CORRELATION ANALYSIS OF CLUSTERS ###
print("\n===== Correlation Analysis of Clusters =====")
clustered_data = df.drop(columns=["Hierarchical_Cluster", "DBSCAN_Cluster"])
correlation_matrix = clustered_data.groupby("KMeans_Cluster").mean().T.corr()

# Plot Correlation Heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Between Clusters", fontsize=14)
plt.show()

print("Correlation Matrix of Clusters:\n", correlation_matrix)

# Save results
df.to_csv("wine_clustering_results.csv", index=False)
```
